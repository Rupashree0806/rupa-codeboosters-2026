```python
import pandas as pd
```

```python
df=pd.read_csv('student_performance.csv')
```

```python
print (df)
```

```python
print("shape:", df.shape)
```

```python
print("Total students:",df.shape[0])
```

```python
print("total columns:",df.shape[1])
```

```python
print("Total departments:", df['department'].nunique())
```

```python
print(df['department'].value_counts())
```

```python
print("Highest Math Score:", df['math_score'].max())
print("lowest math score:",  df['math_score'].min())
print("average math score:", df['math_score'].mean())
print("average Programming score:", df['programming_score'].mean())
print("average attendance", df['attendance_percentage'].mean())
```

```python
# First, create a 'total' score column by summing relevant scores
df['total'] = df['math_score'] + df['programming_score']

print("\n[4] GENDER WISE AVERAGE SCORES")
print("-"*40)
gender_scores=df.groupby('gender')['total'].mean().round(2)
print(gender_scores)
```

```python
print("\n[5] TOP 5 PERFORMERS")
print("-"*40)
top5=df.sort_values('total', ascending=False).head(5)
for index, row in top5.iterrows():
  print(f" {row['name']: <20} | {row['department']:<18} | score:{row['total']}")
```

```python
print("\n[6] STUDNETS WITH LOW ATTNEDANCE(below 75%)")
print("-"*40)
low_att=df[df['attendance_percentage']<75][['name','department','attendance_percentage']]
if len(low_att)>0:
  print(low_att.to_string(index=False))

else:
    print("All students have attendance above 75%")
    print("\n" + "="*60)
    print("Report complete")
    print("="*60)
```
